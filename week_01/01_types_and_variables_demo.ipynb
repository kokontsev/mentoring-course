## Часть 1. Числа

Python динамически типизирован: тип определяется значением. Сейчас посмотрим, как `type()` показывает, что внутри коробки, и как работают три «программистских» операции `//`, `%`, `**`.

```python
x = 42
y = 3.14
name = 'Alice'
print(type(x), type(y), type(name))
```

```python
print(10 / 3)    # 3.333...  обычное деление, всегда float
print(10 // 3)   # 3         целочисленное деление
print(10 % 3)    # 1         остаток
print(2 ** 10)   # 1024      степень
```

А что если сложить два float, которые на бумаге дают ровно `0.3`? Знаменитый подводный камень:

```python
import math

print(0.1 + 0.2)                       # не 0.3
print(0.1 + 0.2 == 0.3)                # False
print(math.isclose(0.1 + 0.2, 0.3))    # True — сравнение с допуском
```

## Часть 2. Строки

Индексация с нуля, срезы `[start:stop:step]`, методы возвращают **новую** строку. Пара `split` / `join` — ежедневный инструмент для CSV и логов. f-строки — современный способ подставить значения в текст.

```python
s = 'Python'
print(s[0])       # 'P'
print(s[-1])      # 'n'   последний символ
print(s[1:4])     # 'yth' срез [start:stop), stop НЕ включается
print(s[::-1])    # 'nohtyP' разворот строки шагом -1
```

```python
text = '  Hello, World!  '
print(text.strip())                       # 'Hello, World!'
print(text.upper())                       # '  HELLO, WORLD!  '
print(text.replace('World', 'Python'))    # '  Hello, Python!  '
print(text)                               # оригинал не изменился
```

```python
csv_row = 'Alice,25,engineer'
parts = csv_row.split(',')
print(parts)                              # ['Alice', '25', 'engineer']
print(' | '.join(parts))                  # 'Alice | 25 | engineer'

name, age, pi = 'Аня', 25, 3.14159
print(f'{name}, вам {age} лет')
print(f'Через 10 лет будет {age + 10}')   # выражения внутри {} работают
print(f'pi ≈ {pi:.2f}')                   # формат: 2 знака после точки
```

## Часть 3. `bool`, `None`, преобразование типов

Внутри Python `bool` наследуется от `int`: `True` это `1`, `False` это `0`. `None` — единственное значение типа `NoneType`, сравнивается через `is`. А `bool('False')` вернёт `True`, потому что строка непустая — содержимое не анализируется.

```python
print(True + True)               # 2 — реально складывается
print(isinstance(True, int))     # True — bool это подкласс int
print(sum([True, False, True]))  # 2 — посчитали True в списке

result = None
print(result is None)            # True — единственный правильный способ
print(type(result))              # <class 'NoneType'>
```

```python
print(int('42'))         # 42
print(str(3.14))         # '3.14'
print(bool(0))           # False
print(bool(''))          # False — пустая строка
print(bool('False'))     # True (!) — непустая строка, неважно что внутри
print(bool('0'))         # True (!) — тот же случай
```

## Часть 4. Переменные — это ссылки

Когда пишешь `x = 42`, Python создаёт объект и связывает имя `x` с ним. Имя — наклейка на коробке, не сама коробка. `b = a` не копия — это второе имя того же объекта. Функция `id(obj)` возвращает уникальный «адрес» объекта в памяти.

```python
a = [1, 2, 3]
b = a                    # b НЕ копия, b — второе имя того же списка
b.append(4)              # меняем через b
print(a)                 # [1, 2, 3, 4] — изменение видно и через a
print(b)                 # [1, 2, 3, 4]

c = [1, 2, 3]            # новый список с тем же содержимым
print(id(a) == id(b))    # True — один объект
print(id(a) == id(c))    # False — разные объекты
```

Отсюда разница: `==` спрашивает «содержимое одинаковое?», `is` — «это буквально один объект в памяти?». Правило: `is` используется **только** для сравнения с `None`.

```python
a = [1, 2, 3]
b = [1, 2, 3]
c = a

print(a == b)      # True — содержимое одинаковое
print(a is b)      # False — разные объекты в памяти
print(a is c)      # True — c это второе имя того же объекта
```

## Часть 5. Изменяемые и неизменяемые типы

Изменяемый объект (`list`, `dict`, `set`) меняется «на месте» — тот же объект, новое содержимое. Неизменяемый (`int`, `float`, `str`, `tuple`) поменять нельзя — любая операция, выглядящая как изменение, создаёт **новый** объект. Покажем через `id()`.

```python
# str — неизменяемый: .upper() возвращает НОВУЮ строку
s = 'привет'
print(id(s))
s = s.upper()
print(id(s))    # другой id! имя s теперь указывает на новый объект

# list — изменяемый: .append() меняет ТОТ ЖЕ объект
numbers = [1, 2, 3]
print(id(numbers))
numbers.append(4)
print(id(numbers))   # тот же id! объект не пересоздан
```

```python
# Если нужна независимая копия — три эквивалентных способа:
items = [1, 2, 3]
backup_a = items.copy()    # явный метод
backup_b = list(items)     # конструктор list
backup_c = items[:]        # пустой срез

items.append(99)
print(items)       # [1, 2, 3, 99]
print(backup_a)    # [1, 2, 3] — независимая копия
print(backup_b)    # [1, 2, 3]
print(backup_c)    # [1, 2, 3]
```
