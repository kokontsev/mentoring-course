## Часть 1. `if` / `elif` / `else`

Сейчас посмотрим, как Python выбирает одну ветку из нескольких. Условия проверяются по порядку сверху вниз: первое истинное — выполняется его блок, остальные пропускаются.

```python
x = 7
if x > 10:
    print('большое')
elif x > 5:
    print('среднее')
else:
    print('маленькое')
```

Заметили? Хотя `x = 7` подходит и под `x > 5`, и под `x > 0`, сработала только первая истинная ветка. Дальше Python даже не проверяет.

```python
# Без elif — ветки независимы и проверяются ВСЕ подряд
score = 85
if score >= 60:
    print('зачёт')
if score >= 80:
    print('отлично')
if score >= 95:
    print('идеал')
# Печатает обе строки 'зачёт' и 'отлично' — это два независимых if
```

```python
# Тернарный оператор — компактная форма для коротких веток
age = 25
status = 'взрослый' if age >= 18 else 'несовершеннолетний'
print(status)
```

```python
# Вложенные условия: if внутри if
user = 'admin'
is_active = True
if user == 'admin':
    if is_active:
        print('доступ открыт')
    else:
        print('админ заблокирован')
else:
    print('обычный пользователь')
```

## Часть 2. Truthy и falsy

В Python в условии можно писать **что угодно**, не только `bool`. Правило: пустое и нулевое — ложь, всё остальное — правда. Сейчас посмотрим, какие значения автоматически считаются ложными.

```python
# Falsy — то, что Python считает ложью в условии
print(bool(0))         # False
print(bool(0.0))       # False
print(bool(''))        # False — пустая строка
print(bool([]))        # False — пустой список
print(bool({}))        # False — пустой словарь
print(bool(None))      # False
```

```python
# Truthy — всё остальное, в том числе подозрительные кейсы
print(bool(1))         # True
print(bool(-5))        # True — отрицательное число тоже True
print(bool('0'))       # True (!) — строка непустая
print(bool('False'))   # True (!) — содержимое не анализируется
print(bool([0]))       # True — список непустой, хоть и с нулём
```

А что если хочется лаконично проверить, что коллекция не пуста? Идиома Python — писать саму коллекцию как условие, без `len() > 0`:

```python
# Идиоматичный Python — короткая проверка
items = [1, 2, 3]
if items:
    print(f'в списке {len(items)} элементов')

name = ''
if not name:
    print('имя не задано')
```

Подводный камень: `if x:` и `if x is not None:` — **разные** проверки. Первая считает falsy и `0`, и `''`, и `[]`. Если переменная может быть `None` или валидным «нулевым» значением — проверяй явно через `is not None`.

```python
# Опасный кейс: 0 — валидное значение, но if x: его отбросит
x = 0
if x:
    print('x задан')
else:
    print('x пустой/нулевой')   # сработает, хотя x = 0 — это валидное число

# Корректная проверка «есть ли вообще значение»
if x is not None:
    print(f'x = {x}')           # сработает: 0 не None
```

## Часть 3. Цикл `for`

`for` перебирает любой итерируемый объект: список, строку, словарь, `range`. Переменная цикла каждый раз принимает значение очередного элемента.

```python
# Перебор списка
fruits = ['яблоко', 'банан', 'вишня']
for fruit in fruits:
    print(fruit)
```

```python
# range(start, stop, step) — генератор чисел. stop НЕ включается
for i in range(5):
    print(i, end=' ')         # 0 1 2 3 4
print()

for i in range(2, 7):
    print(i, end=' ')         # 2 3 4 5 6
print()

for i in range(0, 10, 2):
    print(i, end=' ')         # 0 2 4 6 8
print()
```

```python
# enumerate — пары (индекс, значение). Используй, когда нужен номер
names = ['Аня', 'Борис', 'Вера']
for idx, name in enumerate(names):
    print(idx, name)
```

```python
# zip — параллельный обход нескольких коллекций
names = ['Аня', 'Борис', 'Вера']
grades = [5, 4, 5]
for name, grade in zip(names, grades):
    print(f'{name}: {grade}')
```

```python
# Перебор словаря: .items() даёт пары (ключ, значение)
user = {'name': 'Аня', 'age': 25, 'city': 'Москва'}
for key, value in user.items():
    print(f'{key} -> {value}')
```

А что если коллекции разной длины? `zip` остановится на самой короткой — лишние элементы тихо обрежутся. В Python 3.10+ есть `zip(..., strict=True)`, который поднимет ошибку — защита от опечаток.

```python
# zip обрезает по короткой коллекции
a = [1, 2, 3, 4, 5]
b = ['x', 'y']
for x, y in zip(a, b):
    print(x, y)
# Только две пары: (1, 'x'), (2, 'y'). Хвост 3, 4, 5 потерян
```

## Часть 4. `while`, `break`, `continue`

`while` повторяет блок, пока условие истинно. `break` досрочно выходит из цикла, `continue` пропускает текущую итерацию и переходит к следующей.

```python
# while — цикл по условию
counter = 0
while counter < 5:
    print(counter, end=' ')
    counter += 1            # +=  это сокращение от counter = counter + 1
print()
```

```python
# break — выход из цикла, как только нашли нужное
numbers = [1, 3, 7, 4, 9, 10]
for n in numbers:
    if n % 2 == 0:
        print(f'первое чётное: {n}')
        break
# Печатает 'первое чётное: 4'. Числа 9 и 10 не проверяются
```

```python
# continue — пропустить итерацию, не выходя из цикла
for n in range(10):
    if n % 2 != 0:
        continue            # пропускаем нечётные
    print(n, end=' ')       # 0 2 4 6 8
print()
```

**Задание 2.** Даны два списка: `keys = ['name', 'age', 'city']` и `values = ['Аня', 25, 'Москва']`. Собери из них словарь `{'name': 'Аня', 'age': 25, 'city': 'Москва'}` через цикл.

**Задание 3.** Дан список словарей с заказами. Посчитай суммарную стоимость всех заказов (поле `'total'`).

```python
# Задание 1
nums = [3, 7, 12, 5, 18, 4]
# Твой код:
answ = 'не найдено'
for i in nums:
    if i > 10:
        answ = i
        break
print(answ)
# Ожидаемый вывод: 12
```

```python
# Задание 2
keys = ['name', 'age', 'city']
values = ['Аня', 25, 'Москва']
result = {}
# Твой код:
for key, value in zip(keys, values):
    result[key] = value
result
# Ожидаемый result: {'name': 'Аня', 'age': 25, 'city': 'Москва'}
```

```python
# Задание 3
orders = [
    {'id': 1, 'total': 1200},
    {'id': 2, 'total': 850},
    {'id': 3, 'total': 2100},
    {'id': 4, 'total': 450},
]
total_sum = 0
# Твой код:
for el in orders:
    total_sum += el['total']
print(total_sum)
# Ожидаемый total_sum: 4600
```
