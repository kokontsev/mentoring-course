## Задание 1: Identity vs Equality

Создай два независимых объекта с одинаковым содержимым и третий объект, который ссылается на первый. Покажи на выводе, в каких случаях они сравниваются как равные, а в каких — как один и тот же объект.

Выведи четыре строки:
- сравнение по значению первых двух
- сравнение по идентичности первых двух
- сравнение по значению первого и третьего
- сравнение по идентичности первого и третьего

```python
first = [1, 2, 3]
second = [1, 2, 3]
alias = first

print(f"first == second: {first == second}")
print(f"first is second: {first is second}")
print(f"first == alias:  {first == alias}")
print(f"first is alias:  {first is alias}")
```

**Объясни своими словами:** в каких случаях `==` и `is` дают разный результат, и почему?

`==` сравнивает значения объектов. Два разных списка `[1, 2, 3]` имеют одинаковое содержимое, поэтому `==` возвращает `True`. `is` сравнивает идентичность — то есть, ссылаются ли два имени на один и тот же объект в памяти. `first` и `second` — два разных объекта (созданы независимо), поэтому `is` даёт `False`. `alias = first` создаёт второе имя для того же объекта, поэтому и `==`, и `is` дают `True`.

## Задание 2: Truthy / falsy

В Python почти всё значение можно поместить в условие `if`. Пустые коллекции, ноль и `None` считаются falsy; всё остальное — truthy. Это удобно, но создаёт ловушки.

Возьми значения ниже и для каждого выведи строку вида `<значение> -> True/False`, проверяя его через `bool(...)`.

Значения для проверки: `0`, `0.0`, `""`, `"0"`, `[]`, `[0]`, `{}`, `None`, `False`, `"False"`.

```python
values = [0, 0.0, "", "0", [], [0], {}, None, False, "False"]

for value in values:
    print(f"{value!r:>10} -> {bool(value)}")
```

**Объясни своими словами:** почему `"0"` и `[0]` оказываются truthy, хотя «внутри» у них как будто ничего полезного нет?

Truthy/falsy в Python зависит не от смысла содержимого, а от структуры объекта. `bool(x)` для коллекции возвращает `True`, если в ней есть хотя бы один элемент — не важно, какой. Строка `"0"` — это строка длиной 1, в ней есть символ `0`, поэтому она truthy. Список `[0]` — это список длиной 1, в нём есть элемент (число 0), поэтому он тоже truthy. Падение на этом — частая причина багов: автор кода ожидал, что `if value:` отсеет «нулевые» данные, а отсёк только пустые. Привычка: для числовых проверок пиши `if value != 0`, для пустоты — `if not value` (но помни про `"0"`).

## Задание 3: FizzBuzz

Классическая задача с собеседований. Для чисел от 1 до 15 включительно выводи каждое число на отдельной строке, но с правилами замены:

- если число делится на 3 — выводи `Fizz` вместо числа
- если число делится на 5 — выводи `Buzz`
- если делится и на 3, и на 5 — выводи `FizzBuzz`
- иначе — само число

```python
for n in range(1, 16):
    if n % 15 == 0:
        print("FizzBuzz")
    elif n % 3 == 0:
        print("Fizz")
    elif n % 5 == 0:
        print("Buzz")
    else:
        print(n)
```

## Задание 4: Частота слов

Дан список слов с повторами. Собери словарь, в котором каждому слову сопоставлено число его появлений в списке.

Не используй готовые подсчётчики из стандартной библиотеки — собери словарь вручную через цикл и проверку наличия ключа.

```python
words = ["яблоко", "груша", "яблоко", "банан", "груша", "яблоко"]

frequency = {}
for word in words:
    frequency[word] = frequency.get(word, 0) + 1

print(frequency)
```

**Объясни своими словами:** почему именно `dict` — правильная структура для этой задачи?

Нужно сопоставить каждому уникальному слову число — это пара ключ-значение, ровно то, для чего создан `dict`. Доступ по ключу `O(1)` — даже на миллионе слов проверка «видели ли мы это слово раньше?» работает быстро. Список или множество тут не подходят: список не умеет сопоставлять ключи значениям, а множество не хранит счётчики. `dict.get(word, 0)` удобен тем, что вернёт 0 для нового слова и текущий счётчик для уже встречавшегося — без отдельной ветки `if word in frequency`.

## Задание 5: Сортировка пар по второму элементу

Дан список пар. Отсортируй его по возрастанию второго элемента в каждой паре. Выведи отсортированный список.

Hint: на этой неделе мы ещё не разбирали `lambda` и `key=`. Решение через comprehension + `sort` по умолчанию: построй вспомогательный список из пар вида `(sort_key, original)` — Python сортирует кортежи покомпонентно, поэтому сортировка по первому элементу даст нужный порядок. После сортировки извлеки обратно оригинальные пары.

```python
pairs = [(1, 5), (2, 3), (8, 1), (4, 7), (6, 2)]

# 1. Строим список (sort_key, original) — sort_key это второй элемент каждой пары
keyed = [(p[1], p) for p in pairs]

# 2. Сортируем по умолчанию — кортежи сравниваются по первому элементу
keyed.sort()

# 3. Извлекаем обратно только оригинальные пары через unpacking в comprehension
sorted_pairs = [p for _, p in keyed]

print(sorted_pairs)
```

## Задание 6: Уникальные элементы с сохранением порядка

Дан список с повторами. Собери новый список, в котором каждый элемент встречается только один раз и появляется в том же порядке, в каком он впервые встретился в исходном.

Простой `list(set(items))` не подходит — он теряет порядок. Нужно совместить быструю проверку «видели или нет» с последовательным обходом.

```python
items = ["a", "b", "a", "c", "b", "d", "a", "e", "c"]

seen = set()
result = []
for item in items:
    if item not in seen:
        seen.add(item)
        result.append(item)

print(result)
```

## Задание 7: Проверка анаграмм

Две строки называются анаграммами, если одна получается из другой перестановкой букв (например, `«сон»` и `«нос»`). Проверь три пары и для каждой выведи строку вида `«a» и «b» -> True/False`:

- `«сон»` и `«нос»`
- `«кот»` и `«том»`
- `«листок»` и `«слиток»`

Hint: две строки — анаграммы тогда и только тогда, когда у них совпадают отсортированные списки букв.

```python
pairs = [("сон", "нос"), ("кот", "том"), ("листок", "слиток")]

for a, b in pairs:
    is_anagram = sorted(a) == sorted(b)
    print(f"«{a}» и «{b}» -> {is_anagram}")
```

## Задание 8: Two-sum

Классическая задача с LeetCode. Дан список целых чисел `nums = [2, 7, 11, 15]` и целевое число `target = 9`. Найди индексы **двух разных** элементов, сумма которых равна целевому, и сохрани их в переменную `result` как пару (кортеж).

Гарантировано, что ответ существует и единственен.

Hint: можно решить за два вложенных цикла (перебор всех пар), но есть способ за один проход через словарь — каждое число запоминай вместе с его индексом, и для очередного числа проверяй, лежит ли в словаре нужное дополнение.

```python
nums = [2, 7, 11, 15]
target = 9

# O(n) решение через словарь: число -> его индекс
seen = {}
result = None
for i, num in enumerate(nums):
    complement = target - num
    if complement in seen:
        result = (seen[complement], i)
        break
    seen[num] = i

print(result)
```

## Задание 9: Статистика по оценкам

Дан список оценок учеников: `grades = [4, 5, 3, 5, 2, 4, 5, 3, 4, 5]`.

Посчитай и выведи три значения:
- среднее значение оценок (с округлением до 2 знаков после запятой)
- медиану (середину отсортированного списка)
- процент пятёрок (сколько % всех оценок — это пятёрки, целое число)

Без сторонних библиотек — только встроенные `sum`, `len`, `sorted`, арифметика и/или comprehension.

```python
grades = [4, 5, 3, 5, 2, 4, 5, 3, 4, 5]

# Среднее
average = round(sum(grades) / len(grades), 2)

# Медиана — середина отсортированного списка
ordered = sorted(grades)
middle = len(ordered) // 2
if len(ordered) % 2 == 1:
    median = ordered[middle]
else:
    median = (ordered[middle - 1] + ordered[middle]) / 2

# Процент пятёрок
fives = grades.count(5)
percent_fives = round(fives / len(grades) * 100)

print(f"Среднее:        {average}")
print(f"Медиана:        {median}")
print(f"Процент пятёрок: {percent_fives}%")
```
