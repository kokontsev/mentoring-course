## Задание 3: FizzBuzz

Классическая задача с собеседований. Для чисел от 1 до 15 включительно выводи каждое число на отдельной строке, но с правилами замены:

- если число делится на 3 — выводи `Fizz` вместо числа
- если число делится на 5 — выводи `Buzz`
- если делится и на 3, и на 5 — выводи `FizzBuzz`
- иначе — само число

```python

# TODO: implement
# your code:
for i in range(1, 16):
    if i % 3 == 0 and i % 5 == 0:
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
```

## Задание 4: Частота слов

Дан список слов с повторами. Собери словарь, в котором каждому слову сопоставлено число его появлений в списке.

Не используй готовые подсчётчики из стандартной библиотеки — собери словарь вручную через цикл и проверку наличия ключа.

```python
words = ["яблоко", "груша", "яблоко", "банан", "груша", "яблоко"]

# TODO: implement
# your code:
res = {word: words.count(word) for word in set(words)}
res2 = {}
for word in words:
    if word in res2:
        res2[word] += 1
    else:
        res2[word] = 1
print(res)
print(res2)
```

**Объясни своими словами:** почему именно `dict` — правильная структура для этой задачи?

Из-за проверки на наличие за О(1)

## Задание 5: Сортировка пар по второму элементу

Дан список пар. Отсортируй его по возрастанию второго элемента в каждой паре. Выведи отсортированный список.

Hint: на этой неделе мы ещё не разбирали `lambda` и `key=`. Решение через comprehension + `sort` по умолчанию: построй вспомогательный список из пар вида `(sort_key, original)` — Python сортирует кортежи покомпонентно, поэтому сортировка по первому элементу даст нужный порядок. После сортировки извлеки обратно оригинальные пары.

```python
pairs = [(1, 5), (2, 3), (8, 1), (4, 7), (6, 2)]

# TODO: implement
# your code:
sub = [(x[1], x) for x in pairs]
sub.sort()
sorted_pairs = [x[1] for x in sub]
print(sorted_pairs)
```

## Задание 6: Уникальные элементы с сохранением порядка

Дан список с повторами. Собери новый список, в котором каждый элемент встречается только один раз и появляется в том же порядке, в каком он впервые встретился в исходном.

Простой `list(set(items))` не подходит — он теряет порядок. Нужно совместить быструю проверку «видели или нет» с последовательным обходом.

```python
items = ["a", "b", "a", "c", "b", "d", "a", "e", "c"]

# TODO: implement
# your code:
visit = set()
res = []
for i in items:
    if i not in visit:
        visit.add(i)
        res.append(i)
print(res)
```

## Задание 7: Проверка анаграмм

Две строки называются анаграммами, если одна получается из другой перестановкой букв (например, `«сон»` и `«нос»`). Проверь три пары и для каждой выведи строку вида `«a» и «b» -> True/False`:

## Задание 8: Two-sum

Классическая задача с LeetCode. Дан список целых чисел `nums = [2, 7, 11, 15]` и целевое число `target = 9`. Найди индексы **двух разных** элементов, сумма которых равна целевому, и сохрани их в переменную `result` как пару (кортеж).

Гарантировано, что ответ существует и единственен.

Hint: можно решить за два вложенных цикла (перебор всех пар), но есть способ за один проход через словарь — каждое число запоминай вместе с его индексом, и для очередного числа проверяй, лежит ли в словаре нужное дополнение.

```python
nums = [2, 7, 11, 15]
target = 9

# TODO: implement
# your code:
ost = {}
for i, e in enumerate(nums):
    if target - e in ost:
        print([ost[target - e], i])
        break
    ost[e] = i

```

## Задание 9: Статистика по оценкам

Дан список оценок учеников: `grades = [4, 5, 3, 5, 2, 4, 5, 3, 4, 5]`.

Посчитай и выведи три значения:
- среднее значение оценок (с округлением до 2 знаков после запятой)
- медиану (середину отсортированного списка)
- процент пятёрок (сколько % всех оценок — это пятёрки, целое число)

Без сторонних библиотек — только встроенные `sum`, `len`, `sorted`, арифметика и/или comprehension.


```python
grades = [4, 5, 3, 5, 2, 4, 5, 3, 4, 5]

# TODO: implement
# your code:
print(sum(grades) / len(grades))
if len(grades) % 2 == 1:
    print(sorted(grades)[len(grades) // 2])
else:
    print((sorted(grades)[len(grades) // 2 - 1] + sorted(grades)[len(grades) // 2]) / 2)
print(f'{int(grades.count(5)/len(grades)*100)}%')
```
